```python
import numpy as np
arr = np.arange(1, 21)
arr = arr.reshape(4, 5)
print("Reshaped Array:")
print(arr)
sub_array = arr[1:3, 2:5]
print("\nExtracted Sub-array:")
print(sub_array)
variance = np.var(sub_array)
standard_deviation = np.std(sub_array)
print("\nVariance:")
print(variance)
print("\nStandard Deviation:")
print(standard_deviation)
```
