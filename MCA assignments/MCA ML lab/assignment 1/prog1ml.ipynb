```python
import numpy as np
np.random.seed(10)
arr = np.random.randint(1, 100, size=(10, 10))
print("Original 10 x 10 Array:")
print(arr)
A = arr[0:5, 0:7]
B = arr[0:7, 0:5]
C = arr[5:10, 0:7]
print("\nArray A (5 x 7):")
print(A)
print("\nArray B (7 x 5):")
print(B)
print("\nArray C (5 x 7):")
print(C)
sum_AC = A + C
print("\nSum of A and C:")
print(sum_AC)
multiply_AB = np.dot(A, B)
print("\nMultiplication of A and B:")
print(multiply_AB)
```
