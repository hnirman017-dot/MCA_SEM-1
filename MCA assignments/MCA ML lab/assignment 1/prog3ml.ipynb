```python
import numpy as np
np.random.seed(42)
arr = np.random.randint(10, 100, size=(6, 6))
print("6 x 6 Array:")
print(arr)
row_sum = np.sum(arr, axis=1)
print("\nSum of each row:")
print(row_sum)
column_sum = np.sum(arr, axis=0)
print("\nSum of each column:")
print(column_sum)
row_mean = np.round(np.mean(arr, axis=1), 2)
print("\nMean of each row:")
print(row_mean)
row_max = np.max(arr, axis=1)
print("\nMaximum value in each row:")
print(row_max)
max_index = np.argmax(arr, axis=1)
print("\nColumn index of maximum value in each row:")
print(max_index)
float_arr = arr.astype(float)
print("\nArray converted to float:")
print(float_arr)
```
