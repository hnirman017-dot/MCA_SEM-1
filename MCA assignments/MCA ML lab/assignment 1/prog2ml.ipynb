```python
import numpy as np
np.random.seed(20)
arr = np.random.randint(1, 100, (10, 10))
print("Original 10 x 10 Array:")
print(arr)
print("\n2D Arrays:")
print("\n1 x 100:")
print(arr.reshape(1, 100))
print("\n2 x 50:")
print(arr.reshape(2, 50))
print("\n4 x 25:")
print(arr.reshape(4, 25))
print("\n5 x 20:")
print(arr.reshape(5, 20))
print("\n10 x 10:")
print(arr.reshape(10, 10))
print("\n20 x 5:")
print(arr.reshape(20, 5))
print("\n25 x 4:")
print(arr.reshape(25, 4))
print("\n50 x 2:")
print(arr.reshape(50, 2))
print("\n100 x 1:")
print(arr.reshape(100, 1))
print("\n3D Arrays:")
print("\n1 x 1 x 100:")
print(arr.reshape(1, 1, 100))
print("\n1 x 2 x 50:")
print(arr.reshape(1, 2, 50))
print("\n1 x 4 x 25:")
print(arr.reshape(1, 4, 25))
print("\n1 x 5 x 20:")
print(arr.reshape(1, 5, 20))
print("\n2 x 5 x 10:")
print(arr.reshape(2, 5, 10))
print("\n2 x 10 x 5:")
print(arr.reshape(2, 10, 5))
print("\n4 x 5 x 5:")
print(arr.reshape(4, 5, 5))
print("\n5 x 5 x 4:")
print(arr.reshape(5, 5, 4))
print("\n5 x 10 x 2:")
print(arr.reshape(5, 10, 2))
print("\n10 x 5 x 2:")
print(arr.reshape(10, 5, 2))
```
