```python
import numpy as np
np.random.seed(21)
arr = np.random.uniform(50, 150, 8)
print("Raw Array:")
print(arr)
rounded_arr = np.round(arr, 2)
print("\nRounded Array:")
print(rounded_arr)
zeros = np.zeros((2, 4))
print("\n2 x 4 Array of Zeros:")
print(zeros)
ones = np.ones((2, 4))
print("\n2 x 4 Array of Ones:")
print(ones)
reshaped_arr = rounded_arr.reshape(2, 4)
print("\nReshaped 2 x 4 Array:")
print(reshaped_arr)
result = reshaped_arr + 10
print("\nArray after adding 10:")
print(result)
```
